```python
import ipywidgets as widgets
from IPython.display import display, clear_output

all_quizzes = {
    "19_1_1: Is It Any Good?": [
        {
            "question": "predicted and breed are NumPy arrays of the same length. What does (predicted == breed).mean() give?",
            "options": [
                "The accuracy: the fraction of positions where the prediction matches the true breed",
                "The most common breed",
                "An error, because arrays can't be compared",
                "The number of right predictions"
            ],
            "answer": "The accuracy: the fraction of positions where the prediction matches the true breed"
        },
        {
            "question": "The four breeds are balanced in the test set. What accuracy does always predicting the most common breed get?",
            "options": [
                "0.50",
                "1.00",
                "0.90",
                "0.25"
            ],
            "answer": "0.25"
        },
        {
            "question": "If 90 of every 100 photos were beagles, why would an accuracy of 0.90 be unimpressive?",
            "options": [
                "The model would have to be retrained",
                "Accuracy can't be computed on unbalanced data",
                "Beagles are the easiest breed to recognize",
                "Always answering \"beagle\" would also score 0.90 without looking at the photos"
            ],
            "answer": "Always answering \"beagle\" would also score 0.90 without looking at the photos"
        },
        {
            "question": "Why is a model's accuracy on its own training photos not a fair measure?",
            "options": [
                "The model was trained to get those photos right, so it can learn details that don't carry over to new photos",
                "Training photos are never labeled",
                "Training photos are lower quality than test photos",
                "Training accuracy is always exactly 1.0"
            ],
            "answer": "The model was trained to get those photos right, so it can learn details that don't carry over to new photos"
        },
        {
            "question": "During 300 epochs of training, training accuracy reached 1.0 while validation loss started to rise after about epoch 32. What is this called?",
            "options": [
                "A baseline",
                "Softmax",
                "The accuracy paradox",
                "Overfitting"
            ],
            "answer": "Overfitting"
        },
        {
            "question": "Thirty versions of the model were compared on the validation photos and the best one was chosen. Why is its validation score optimistic?",
            "options": [
                "It was trained on the validation photos",
                "It was chosen for having the highest validation score, so part of its lead is luck that won't carry over to new photos",
                "The validation score is always rounded up",
                "Validation photos are easier than test photos"
            ],
            "answer": "It was chosen for having the highest validation score, so part of its lead is luck that won't carry over to new photos"
        }
    ],
    "19_1_2: The Confusion Matrix": [
        {
            "question": "In the Pets confusion matrix, rows are true breeds and columns are predicted breeds. What does the cell in the Ragdoll row and Birman column (12) mean?",
            "options": [
                "The model said Birman 12 times in total",
                "12 real Birmans were predicted to be Ragdolls",
                "12 real Ragdolls were predicted to be Birmans",
                "12 Ragdolls were predicted correctly"
            ],
            "answer": "12 real Ragdolls were predicted to be Birmans"
        },
        {
            "question": "Where in a confusion matrix are the right answers?",
            "options": [
                "Off the diagonal",
                "On the diagonal",
                "In the first row",
                "In the last column"
            ],
            "answer": "On the diagonal"
        },
        {
            "question": "What does Ragdoll recall measure?",
            "options": [
                "The fraction of all photos that are Ragdolls",
                "The model's overall accuracy",
                "When the model says Ragdoll, the fraction of the time it is right",
                "Of the real Ragdolls, the fraction the model found"
            ],
            "answer": "Of the real Ragdolls, the fraction the model found"
        },
        {
            "question": "12 Ragdolls were called Birmans. Which two numbers does that mistake lower?",
            "options": [
                "Ragdoll's recall and Birman's precision",
                "Birman's recall and Ragdoll's precision",
                "Only the overall accuracy",
                "Ragdoll's precision and Birman's recall"
            ],
            "answer": "Ragdoll's recall and Birman's precision"
        },
        {
            "question": "On a set with 5 Ragdolls and 150 other photos, a detector that never says \"Ragdoll\" has accuracy 0.968. What does this show?",
            "options": [
                "The detector is as useful as the model",
                "With a rare class, accuracy can look high for a model that never finds that class",
                "Accuracy is always the best metric",
                "Ragdolls are impossible to detect"
            ],
            "answer": "With a rare class, accuracy can look high for a model that never finds that class"
        },
        {
            "question": "Lowering the threshold for saying \"Ragdoll\" from 0.5 to 0.2 usually does what?",
            "options": [
                "Raises recall and lowers precision",
                "Raises both precision and recall",
                "Raises precision and lowers recall",
                "Changes nothing, because the model is the same"
            ],
            "answer": "Raises recall and lowers precision"
        }
    ],
    "19_1_3: One Neuron": [
        {
            "question": "x = [2, 0.5, 1], w = [1.5, -2, 0.25], b = -0.5. What is x @ w + b?",
            "options": [
                "1.75",
                "2.25",
                "3.25",
                "-0.5"
            ],
            "answer": "1.75"
        },
        {
            "question": "What does logistic regression do to turn its weighted sum z into a probability?",
            "options": [
                "Divides z by the number of features",
                "Takes the absolute value of z",
                "Applies sigmoid, 1 / (1 + e^(-z))",
                "Rounds z to 0 or 1"
            ],
            "answer": "Applies sigmoid, 1 / (1 + e^(-z))"
        },
        {
            "question": "What is sigmoid(0)?",
            "options": [
                "1",
                "0",
                "e",
                "0.5"
            ],
            "answer": "0.5"
        },
        {
            "question": "The Pets model's last layer has a weight matrix W of shape (4, 512). What does each row hold?",
            "options": [
                "The 512 weights for one breed",
                "The 512 features of one photo",
                "The biases for the four breeds",
                "The 4 scores for one photo"
            ],
            "answer": "The 512 weights for one breed"
        },
        {
            "question": "features has shape (800, 512) and W.T has shape (512, 4). What is the shape of features @ W.T?",
            "options": [
                "(512, 512)",
                "An error, because the shapes don't match",
                "(800, 4)",
                "(4, 800)"
            ],
            "answer": "(800, 4)"
        },
        {
            "question": "Why does features @ W raise an error when W has shape (4, 512)?",
            "options": [
                "You can't multiply more than 4 rows at once",
                "The inner sizes (512 and 4) don't match",
                "features must have shape (4, 512)",
                "W must be transposed twice"
            ],
            "answer": "The inner sizes (512 and 4) don't match"
        }
    ],
    "19_1_4: Softmax and XOR": [
        {
            "question": "What are the two steps of softmax?",
            "options": [
                "Sort the scores, then keep the largest",
                "Divide each score by the largest score, then round",
                "Add 1 to each score, then take the log",
                "Raise e to the power of each score, then divide each result by the total"
            ],
            "answer": "Raise e to the power of each score, then divide each result by the total"
        },
        {
            "question": "Adding 10 to all four scores before softmax does what to the probabilities?",
            "options": [
                "Makes the probabilities add up to more than 1",
                "Makes all four probabilities equal",
                "Makes the top probability larger",
                "Nothing; only the differences between scores matter"
            ],
            "answer": "Nothing; only the differences between scores matter"
        },
        {
            "question": "The model gave the true breed a probability of 0.01. What is that photo's cross-entropy penalty, -log(0.01)?",
            "options": [
                "About 0.01",
                "0",
                "About 0.69",
                "About 4.6"
            ],
            "answer": "About 4.6"
        },
        {
            "question": "Why can't a single neuron solve XOR?",
            "options": [
                "The XOR points have no labels",
                "Its decision boundary is a straight line, and no straight line separates the two 1s from the two 0s",
                "Sigmoid can't output values above 0.5",
                "XOR has too few points to train on"
            ],
            "answer": "Its decision boundary is a straight line, and no straight line separates the two 1s from the two 0s"
        },
        {
            "question": "What did logistic regression do on the four XOR points?",
            "options": [
                "Failed to run because XOR has only two features",
                "Got all four points right",
                "Got three of the four points right with confident probabilities",
                "Set both weights to 0 and gave every point a probability of 0.5"
            ],
            "answer": "Set both weights to 0 and gave every point a probability of 0.5"
        },
        {
            "question": "In the hand-built XOR network, what do the two hidden neurons compute?",
            "options": [
                "The two inputs, unchanged",
                "The softmax of the two inputs",
                "Whether at least one input is 1, and whether both inputs are 1",
                "The final answer, twice"
            ],
            "answer": "Whether at least one input is 1, and whether both inputs are 1"
        }
    ]
}

class QuizApp:
    def __init__(self, quiz_name, data):
        self.quiz_name = quiz_name
        self.data = data
        self.current_index = 0
        self.score = 0

        self.header = widgets.HTML(value="<h2>" + quiz_name + "</h2>")
        self.progress = widgets.HTML(value="")
        self.question_text = widgets.HTML(value="")
        self.options = widgets.RadioButtons(options=[], layout={"width": "max-content"})
        self.submit_btn = widgets.Button(description="Submit Answer", button_style="primary")
        self.next_btn = widgets.Button(description="Next Question", button_style="info", disabled=True)
        self.back_btn = widgets.Button(description="Back to Menu", button_style="warning")
        self.feedback = widgets.HTML(value="")

        self.submit_btn.on_click(self.handle_submit)
        self.next_btn.on_click(self.handle_next)
        self.back_btn.on_click(self.handle_back)
        self.render_question()

    def render_question(self):
        clear_output(wait=True)
        q = self.data[self.current_index]
        self.progress.value = "<b>Question " + str(self.current_index + 1) + " of " + str(len(self.data)) + "</b>"
        self.question_text.value = "<p style='font-size:14px;'>" + q["question"] + "</p>"
        self.options.options = q["options"]
        self.options.value = None
        self.submit_btn.disabled = False
        self.next_btn.disabled = True
        self.feedback.value = ""
        display(self.header, self.progress, self.question_text, self.options,
                widgets.HBox([self.submit_btn, self.next_btn]), self.feedback)

    def handle_submit(self, b):
        if self.options.value is None:
            self.feedback.value = "<span style='color:red;'>Please select an option!</span>"
            return
        correct = self.data[self.current_index]["answer"]
        if self.options.value == correct:
            self.score += 1
            self.feedback.value = "<span style='color:green; font-weight:bold;'>Correct!</span>"
        else:
            self.feedback.value = "<span style='color:red; font-weight:bold;'>Incorrect.</span><br>The correct answer was: <b>" + correct + "</b>"
        self.submit_btn.disabled = True
        self.next_btn.disabled = False

    def handle_next(self, b):
        self.current_index += 1
        if self.current_index < len(self.data):
            self.render_question()
        else:
            self.show_results()

    def handle_back(self, b):
        show_menu()

    def show_results(self):
        clear_output()
        pct = (self.score / len(self.data)) * 100
        html = """<div style="border: 2px solid #0366d6; padding: 20px; border-radius: 10px;">
            <h2>Quiz Complete!</h2>
            <p style="font-size:16px;">Your Final Score: <b>""" + str(self.score) + " / " + str(len(self.data)) + """</b></p>
            <p style="font-size:20px; color:#0366d6;"><b>Percentage: """ + "{:.1f}".format(pct) + "%</b></p></div>"
        display(widgets.HTML(html), self.back_btn)

def show_menu():
    clear_output(wait=True)
    header = widgets.HTML(value="<h2>19_1 The On-Ramp: Practice Quizzes</h2>")
    instruction = widgets.HTML(value="<p>Select a notebook quiz below to begin.</p>")
    quiz_selector = widgets.Dropdown(options=list(all_quizzes.keys()), description="Select Quiz:", layout={"width": "50%"})
    start_btn = widgets.Button(description="Start Quiz", button_style="success")

    def on_start(b):
        QuizApp(quiz_selector.value, all_quizzes[quiz_selector.value])

    start_btn.on_click(on_start)
    display(header, instruction, quiz_selector, start_btn)

show_menu()
```
