```python
import ipywidgets as widgets
from IPython.display import display, clear_output

all_quizzes = {
    "19_0_1: Your First Image Model": [
        {
            "question": "Why does 19_0_1 have you train a working image model before explaining how neural networks work?",
            "options": [
                "Neural networks can't be explained without first training one",
                "Later units use a different kind of model, so this one doesn't need explaining",
                "The training code is too short to need an explanation",
                "The course shows the whole model working first and then explains its parts in later units"
            ],
            "answer": "The course shows the whole model working first and then explains its parts in later units"
        },
        {
            "question": "What are the validation photos used for in 19_0_1?",
            "options": [
                "Checking the model's accuracy after each epoch on photos it never learns from",
                "One final check at the very end, after all decisions are made",
                "Replacing photos that failed to load",
                "Training the new last layer"
            ],
            "answer": "Checking the model's accuracy after each epoch on photos it never learns from"
        },
        {
            "question": "What does it mean that ResNet-18 is pretrained?",
            "options": [
                "It was trained on our four pet breeds before class",
                "Its weights were already learned on ImageNet before we downloaded it",
                "It can only be used for the 1,000 ImageNet categories",
                "It has no weights until we train it"
            ],
            "answer": "Its weights were already learned on ImageNet before we downloaded it"
        },
        {
            "question": "Why does the pretrained model call the Birman a \"Siamese cat\" before fine-tuning?",
            "options": [
                "Siamese is the most common category in ImageNet",
                "The photo was mislabeled in the dataset",
                "ImageNet has no Birman category, so it picks the closest category it has",
                "The model was trained mostly on dogs"
            ],
            "answer": "ImageNet has no Birman category, so it picks the closest category it has"
        },
        {
            "question": "In 19_0_1, which part of ResNet-18 does training change?",
            "options": [
                "Only the new last layer (2,052 numbers); every other weight is frozen",
                "Every layer except the new last layer",
                "All of its roughly 11.7 million weights",
                "None of it; training only changes the photos"
            ],
            "answer": "Only the new last layer (2,052 numbers); every other weight is frozen"
        },
        {
            "question": "What is an epoch?",
            "options": [
                "The time it takes to download the pretrained weights",
                "One prediction on a single photo",
                "One batch of 32 photos",
                "One pass through all of the training photos"
            ],
            "answer": "One pass through all of the training photos"
        },
        {
            "question": "Why does fine-tuning work with only 100 training photos per breed?",
            "options": [
                "100 photos is enough to train any network from scratch",
                "The four breeds are very easy to tell apart",
                "The test photos are also used for training",
                "The pretrained network already learned what animals look like from over a million ImageNet photos"
            ],
            "answer": "The pretrained network already learned what animals look like from over a million ImageNet photos"
        },
        {
            "question": "You upload a photo of a mixed-breed dog. What will the model do?",
            "options": [
                "Say that the dog isn't one of the four breeds",
                "Refuse to make a prediction",
                "Add a fifth breed and retrain itself",
                "Answer with one of its four breeds, because those are the only answers it has"
            ],
            "answer": "Answer with one of its four breeds, because those are the only answers it has"
        }
    ],
    "19_0_2: What Just Happened?": [
        {
            "question": "Why does 19_0_2 load the course copy of the last layer instead of each student's own?",
            "options": [
                "The course copy was trained on more photos",
                "So everyone works with the same model, whose numbers 19_1 and the unit 19 exam refer to",
                "Students' own models can't be saved in Colab",
                "The course copy has more than four breeds"
            ],
            "answer": "So everyone works with the same model, whose numbers 19_1 and the unit 19 exam refer to"
        },
        {
            "question": "What are a photo's 512 features in ResNet-18?",
            "options": [
                "Named measurements such as ear length and fur color",
                "The 512 photos in the training set",
                "The photo's 512 brightest pixels",
                "The numbers the rest of the network computes from the photo and passes to the last layer"
            ],
            "answer": "The numbers the rest of the network computes from the photo and passes to the last layer"
        },
        {
            "question": "What does softmax do to the model's four scores?",
            "options": [
                "Rounds each score to 0 or 1",
                "Picks the lowest score",
                "Adds the four scores together",
                "Turns them into four positive probabilities that add up to 1"
            ],
            "answer": "Turns them into four positive probabilities that add up to 1"
        },
        {
            "question": "In the table of true breed against predicted breed, where are the right answers?",
            "options": [
                "On the diagonal, from top left to bottom right",
                "In the last column",
                "In the first row",
                "Everywhere except the diagonal"
            ],
            "answer": "On the diagonal, from top left to bottom right"
        },
        {
            "question": "Which kind of mistake did the model make most often on the test photos?",
            "options": [
                "Calling a Ragdoll a Birman",
                "Calling a dog a cat",
                "Calling a beagle a basset hound",
                "Calling a cat a dog"
            ],
            "answer": "Calling a Ragdoll a Birman"
        },
        {
            "question": "With four breeds, why can the model's confidence never be below 0.25?",
            "options": [
                "Softmax adds 0.25 to every probability",
                "The highest of four probabilities that add up to 1 must be at least one quarter",
                "Random guessing is always right 25% of the time",
                "The model rounds small probabilities up to 0.25"
            ],
            "answer": "The highest of four probabilities that add up to 1 must be at least one quarter"
        },
        {
            "question": "What did the confidence plot show about right and wrong answers?",
            "options": [
                "Wrong answers were more confident than right ones",
                "Wrong answers were less confident on average, but a few still had confidence above 0.8",
                "Right and wrong answers had exactly the same confidence",
                "Every wrong answer had confidence below 0.5"
            ],
            "answer": "Wrong answers were less confident on average, but a few still had confidence above 0.8"
        },
        {
            "question": "What happened when the model was shown breeds it was never trained on?",
            "options": [
                "Every photo got a low confidence close to 0.25",
                "The model correctly said none of the four breeds",
                "Each photo got one of the four breeds, some very confidently, and some dogs got cat breeds",
                "The model refused to predict"
            ],
            "answer": "Each photo got one of the four breeds, some very confidently, and some dogs got cat breeds"
        }
    ]
}

class QuizApp:
    def __init__(self, quiz_name, data):
        self.quiz_name = quiz_name
        self.data = data
        self.current_index = 0
        self.score = 0

        self.header = widgets.HTML(value="<h2>" + quiz_name + "</h2>")
        self.progress = widgets.HTML(value="")
        self.question_text = widgets.HTML(value="")
        self.options = widgets.RadioButtons(options=[], layout={"width": "max-content"})
        self.submit_btn = widgets.Button(description="Submit Answer", button_style="primary")
        self.next_btn = widgets.Button(description="Next Question", button_style="info", disabled=True)
        self.back_btn = widgets.Button(description="Back to Menu", button_style="warning")
        self.feedback = widgets.HTML(value="")

        self.submit_btn.on_click(self.handle_submit)
        self.next_btn.on_click(self.handle_next)
        self.back_btn.on_click(self.handle_back)
        self.render_question()

    def render_question(self):
        clear_output(wait=True)
        q = self.data[self.current_index]
        self.progress.value = "<b>Question " + str(self.current_index + 1) + " of " + str(len(self.data)) + "</b>"
        self.question_text.value = "<p style='font-size:14px;'>" + q["question"] + "</p>"
        self.options.options = q["options"]
        self.options.value = None
        self.submit_btn.disabled = False
        self.next_btn.disabled = True
        self.feedback.value = ""
        display(self.header, self.progress, self.question_text, self.options,
                widgets.HBox([self.submit_btn, self.next_btn]), self.feedback)

    def handle_submit(self, b):
        if self.options.value is None:
            self.feedback.value = "<span style='color:red;'>Please select an option!</span>"
            return
        correct = self.data[self.current_index]["answer"]
        if self.options.value == correct:
            self.score += 1
            self.feedback.value = "<span style='color:green; font-weight:bold;'>Correct!</span>"
        else:
            self.feedback.value = "<span style='color:red; font-weight:bold;'>Incorrect.</span><br>The correct answer was: <b>" + correct + "</b>"
        self.submit_btn.disabled = True
        self.next_btn.disabled = False

    def handle_next(self, b):
        self.current_index += 1
        if self.current_index < len(self.data):
            self.render_question()
        else:
            self.show_results()

    def handle_back(self, b):
        show_menu()

    def show_results(self):
        clear_output()
        pct = (self.score / len(self.data)) * 100
        html = """<div style="border: 2px solid #0366d6; padding: 20px; border-radius: 10px;">
            <h2>Quiz Complete!</h2>
            <p style="font-size:16px;">Your Final Score: <b>""" + str(self.score) + " / " + str(len(self.data)) + """</b></p>
            <p style="font-size:20px; color:#0366d6;"><b>Percentage: """ + "{:.1f}".format(pct) + "%</b></p></div>"
        display(widgets.HTML(html), self.back_btn)

def show_menu():
    clear_output(wait=True)
    header = widgets.HTML(value="<h2>19_0 The Whole Game: Practice Quizzes</h2>")
    instruction = widgets.HTML(value="<p>Select a notebook quiz below to begin.</p>")
    quiz_selector = widgets.Dropdown(options=list(all_quizzes.keys()), description="Select Quiz:", layout={"width": "50%"})
    start_btn = widgets.Button(description="Start Quiz", button_style="success")

    def on_start(b):
        QuizApp(quiz_selector.value, all_quizzes[quiz_selector.value])

    start_btn.on_click(on_start)
    display(header, instruction, quiz_selector, start_btn)

show_menu()
```
